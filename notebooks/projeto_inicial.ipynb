```python
import polars as pl

dados = pl.read_csv(r"C:\Users\Gabriel Barbosa\Downloads\script_job_6d1db9d69b9435cf5ce2401e037888fb_1.csv", 
                    separator=",", 
                    encoding="utf8",
                    quote_char='"',
                    infer_schema_length=0)

df = pl.DataFrame(dados)

df.head(3)
```

```python
dim_emenda = (
    df
    .filter(
    pl.col('codigo_da_emenda')!='Sem informação'
    )
    .select([
    # Identificador da emenda
        pl.col("codigo_da_emenda").cast(pl.Int64).alias("codigo_da_emenda"),
        pl.col("tipo_de_emenda").str.strip_chars().alias("tipo_emenda"),
        pl.col("ano_da_emenda").cast(pl.Int64).alias("ano_emenda"),
        pl.col("numero_da_emenda").cast(pl.Int64).alias("numero_emenda"),
        
        # Autor
        pl.col("codigo_do_autor_da_emenda").cast(pl.Int64).alias("codigo_autor"),
        pl.col("nome_do_autor_da_emenda").str.strip_chars().alias("nome_autor"),
        pl.col("possui_apoiador_solicitante").str.strip_chars().alias("possui_apoiador_solicitante"),
        
        # Classificação Orçamentária
        pl.col("codigo_funcao").replace("MU", None).cast(pl.Int64).alias("codigo_funcao"),
        pl.col("nome_funcao").str.strip_chars().alias("nome_funcao"),
        pl.col("codigo_subfuncao").replace("MU", None).cast(pl.Int64).alias("codigo_subfuncao"),
        pl.col("nome_subfuncao").str.strip_chars().alias("nome_subfuncao"),

        pl.col('codigo_programa').replace('MU', None).cast(pl.Int64).alias('codigo_programa'),
        pl.col('nome_programa').str.strip_chars().alias('nome_programa'),
        pl.col('codigo_acao').str.strip_chars().alias('codigo_acao'),
        pl.col('nome_acao').str.strip_chars().alias('nome_acao'),

        pl.col('codigo_plano_orcamentario').str.strip_chars().alias('codigo_plano_orcamentario'),
        pl.col('nome_plano_orcamentario').str.strip_chars().alias('nome_plano_orcamentario'),

        # Localidade principal
        pl.col("uf").str.strip_chars().alias("Estado"),

        pl.when(pl.col("uf") == "ACRE").then(pl.lit("AC"))
            .when(pl.col("uf") == "ALAGOAS").then(pl.lit("AL"))
            .when(pl.col("uf") == "AMAPÁ").then(pl.lit("AP"))
            .when(pl.col("uf") == "AMAZONAS").then(pl.lit("AM"))
            .when(pl.col("uf") == "BAHIA").then(pl.lit("BA"))
            .when(pl.col("uf") == "CEARÁ").then(pl.lit("CE"))
            .when(pl.col("uf") == "DISTRITO FEDERAL").then(pl.lit("DF"))
            .when(pl.col("uf") == "ESPÍRITO SANTO").then(pl.lit("ES"))
            .when(pl.col("uf") == "GOIÁS").then(pl.lit("GO"))
            .when(pl.col("uf") == "MARANHÃO").then(pl.lit("MA"))
            .when(pl.col("uf") == "MATO GROSSO").then(pl.lit("MT"))
            .when(pl.col("uf") == "MATO GROSSO DO SUL").then(pl.lit("MS"))
            .when(pl.col("uf") == "MINAS GERAIS").then(pl.lit("MG"))
            .when(pl.col("uf") == "PARÁ").then(pl.lit("PA"))
            .when(pl.col("uf") == "PARAÍBA").then(pl.lit("PB"))
            .when(pl.col("uf") == "PARANÁ").then(pl.lit("PR"))
            .when(pl.col("uf") == "PERNAMBUCO").then(pl.lit("PE"))
            .when(pl.col("uf") == "PIAUÍ").then(pl.lit("PI"))
            .when(pl.col("uf") == "RIO DE JANEIRO").then(pl.lit("RJ"))
            .when(pl.col("uf") == "RIO GRANDE DO NORTE").then(pl.lit("RN"))
            .when(pl.col("uf") == "RIO GRANDE DO SUL").then(pl.lit("RS"))
            .when(pl.col("uf") == "RONDÔNIA").then(pl.lit("RO"))
            .when(pl.col("uf") == "RORAIMA").then(pl.lit("RR"))
            .when(pl.col("uf") == "SANTA CATARINA").then(pl.lit("SC"))
            .when(pl.col("uf") == "SÃO PAULO").then(pl.lit("SP"))
            .when(pl.col("uf") == "SERGIPE").then(pl.lit("SE"))
            .when(pl.col("uf") == "TOCANTINS").then(pl.lit("TO"))
            .when(pl.col("uf") == "Múltiplo").then(pl.lit("MULT"))
            .when(pl.col("uf") == "Sem informação").then(pl.lit("NACI"))
            .otherwise(pl.col("uf"))
            .alias("uf"),
        
        pl.col("localidade_de_aplicacao_do_recurso").str.strip_chars().alias("localidade_recurso"),
        pl.col("codigo_uf_ibge").cast(pl.Int64).alias("codigo_uf_ibge"),
        pl.col("regiao").str.strip_chars().alias("regiao"),

        pl.col("codigo_municipio_ibge").alias("codigo_muninini_ibge"),
        pl.when(
            pl.col('uf') == "Sem informação")
            .then (pl.lit(-1))
                .when(pl.col('uf') == "Múltiplo")
                .then (pl.lit(-14))
            .otherwise(pl.col('codigo_municipio_ibge').replace("Sem informação", -1).cast(pl.Int64))
        .alias('codigo_municipio_ibge'),


        pl.col("municipio"). alias('munini'),
        pl.when(
            pl.col("municipio").replace("Sem informação", None).is_null()
            & pl.col("localidade_de_aplicacao_do_recurso").str.strip_chars().str.ends_with("Nacional"))
            .then(pl.lit("Nacional"))
            .when(
                pl.col("municipio").replace("Sem informação", None).is_null()
                & pl.col("localidade_de_aplicacao_do_recurso").str.strip_chars().str.ends_with("(UF)"))
                .then(pl.lit("Estadual"))
                    .when(
                        pl.col("municipio") == "Sem informação")
                        .then(pl.col('regiao').str.strip_chars())
            .otherwise(
        pl.col("municipio").str.strip_chars())
        .alias("municipio")
    ])
    .unique(subset=['codigo_da_emenda'])
    ).sort(by='ano_emenda', descending=True)


dim_emenda.write_csv(r"C:\Users\Gabriel Barbosa\Downloads\dim_emenda2.csv", include_header=True, separator=',', )
print(dim_emenda)
```

```python
import polars as pl

dados_c = pl.read_csv(r"C:\Users\Gabriel Barbosa\Downloads\job_JRrPmEMITywCw-1PMK3MrFXPK_H9.csv", 
                    separator=",", 
                    encoding="utf8",
                    quote_char='"',
                    infer_schema_length=0)

df_c = pl.DataFrame(dados_c)

df_c.head(3)
```

```python

fato_convenios = (
    df_c
    .filter(
    pl.col("codigo_da_emenda").is_not_null()
    & (pl.col('codigo_da_emenda')!='Sem informação')
    )
    .select([
        pl.col("codigo_da_emenda").cast(pl.Int64).alias("codigo_da_emenda"),
        pl.col('codigo_funcao').cast(pl.Int64).alias('codigo_funcao'),
        pl.col('codigo_subfuncao').cast(pl.Int64).alias('codigo_subfuncao'),
        pl.col("numero_convenio").alias("numero_convenio"),
        pl.col('data_publicacao_convenio').str.strptime(pl.Date, format="%d/%m/%Y").alias('data_publicacao_convenio'),

        pl.col('nome_funcao').alias('nome_funcao'),
        pl.col("nome_subfuncao").alias("nome_subfuncao"),

        pl.col("localidade_do_gasto").alias("localidade_do_gasto"),

        pl.when(
            pl.col("localidade_do_gasto").str.strip_chars().str.ends_with("Nacional"))
            .then(pl.lit("Nacional"))
                .when(
                pl.col("localidade_do_gasto").str.strip_chars().str.ends_with("(UF)"))
                .then(pl.lit("Estadual"))
                    .when(
                        pl.col("localidade_do_gasto").str.strip_chars().str.contains(r"(Norte|Sul|Centro-Oeste|Nordeste|Sudeste)$"))
                        .then(pl.lit('Regional'))
                .otherwise(pl.col("localidade_do_gasto").str.strip_chars().str.slice(-2))
                .alias("localidade_classificada"),

        pl.col("tipo_de_emenda").alias("tipo_de_emenda"),
        pl.col("convenente").alias("convenente"),
        pl.col("objeto_convenio").alias("objeto_convenio"),
        pl.col("valor_convenio").str.replace_all(r"\.", "").str.replace(",", ".").cast(pl.Float64, strict=False).alias("valor_convenio"),
    
    ])
    .unique(subset=['numero_convenio', 'codigo_da_emenda'])
    ).sort(by='data_publicacao_convenio', descending=True)


fato_convenios.write_csv(r"C:\Users\Gabriel Barbosa\Downloads\fato_convenios.csv", include_header=True, separator=',', )
print(fato_convenios)
```
