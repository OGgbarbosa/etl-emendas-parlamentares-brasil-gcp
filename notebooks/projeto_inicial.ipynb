```python
from pathlib import Path

import polars as pl

DATABASE_DIR = Path("../database") if Path("../database").exists() else Path("../database_sample")
EMENDAS_CSV = DATABASE_DIR / "EmendasParlamentares.csv"
CONVENIOS_CSV = DATABASE_DIR / "EmendasParlamentares_Convenios.csv"
FAVORECIDOS_CSV = DATABASE_DIR / "EmendasParlamentares_PorFavorecido.csv"


with open(EMENDAS_CSV, "r", encoding="latin1") as f:
    conteudo = f.read()

EMENDAS_UTF8 = EMENDAS_CSV.with_name(f"{EMENDAS_CSV.stem}_utf8.csv")

with open(EMENDAS_UTF8, "w", encoding="utf-8", newline="") as f:
    f.write(conteudo)


emendas = pl.scan_csv(EMENDAS_UTF8, separator=";", encoding="utf8")
# convenios = pl.scan_csv(CONVENIOS_CSV, separator=';', encoding='latin1')
# favorecidos = pl.scan_csv(FAVORECIDOS_CSV, separator=';', encoding='latin1')

df_emenda = emendas
# df_convenio = convenios
# df_favorecidos = favorecidos
```
